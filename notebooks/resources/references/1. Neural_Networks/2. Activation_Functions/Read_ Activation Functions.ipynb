So briefly until this portion of the lecture, we have come to understand why we need an activation function. In order to create a nonlinear decision boundary, activation function plays a vital role.
Now, let's discuss what an activation function really is.

The input to an activation function is the weighted sum (between the weights and inputs to that particular neuron), also called as $Z$ value. Now this value of $Z$ can vary from $[-\infty$  to $+ \infty] $ which is a large number of values to work on. **Inspired from the concept of neurons firing only after the stimulus reaching a certain threshold,** the activation function here, decides **whether the neuron should fire or not based on the values of $Z$**. This concept works fine when we are talking about classification based on two classes. (_For a linearly Separable data_)

Algorithmically,   

$ If(Activation Function >= Threshold Value)$   
$\space\space\space\space\space$Fire/Activate the Neuron   
else   
$\space\space\space\space\space$Deactivate the Neuron

**Problem formulation**   
What if we have a multiclass classification problem like classification of  fruits in a basket containing more than 2 classes of fruits, say 4?    
Our **step activated or Binary (1/0) or (Yes/No) or digital activation** neuron gets activated for more than one class of fruits making it impossible for us to distinguish which fruit. For example: It might get activated for apple along with pear and orange but deactivated for banana and mango at the same time.   
**So what we need is an activation function that outputs say, 40% of Class 1, 30% of class 2 and so on rather than 100% of Class 1 and 2 and 3**. This concept is called as **analog activation** (for example giving a probability from 0 to 1, analog values).

Similarly, a neural network works on the concept of learning parameters based on optimization algorithms like Gradient Descent. considering a **linear activation function of form  $y=cx$, the derivative of this function is a constant $c$ which doesnt propagate errors or update the learning parameters.**

Hence, **an ideal activation function will output analog activations and easily propagate errors** i.e. is differential with respect to the learning parameters.

## **Types of Activation Functions**

### **Sigmoid Activation Function**   
The input-output mapping function in sigmoid activation is $\phi(z)=\frac{1}{1+e^{-z}}$ 

This function mostly **appears in the output layer of DL architecture and is used for probability based output and has been applied successfully in binary classification problem and logistic regression.**

Sigmoid is a non-linear activation function used mostly in feed forward neural networks. It is a bounded differentiable real funciton defined for real input values with positive derivatives everywhere and some degree of smoothness.
The figure below shows the plot of mapping between input and output:

**The above curve shows that any input will be scaled to a value between [0 to 1] for any input ranging from $- \infty \space to + \infty $**
$$ 0< \phi(z) < 1$$

**Domain** :$- \infty \space to + \infty $   
**Range** :[$0 \space to 1$]

For example,

When x=2, $\rightarrow f(z)=\frac{1}{1+e^{-2}}=0.8808$

When x=-1,$\rightarrow f(z)=\frac{1}{1+e^1}=0.26894$

Similarly, The derivative of sigmoid activation function, $y$ with respect to input values $z$ is computed as:

$$
\begin{align*}
y &= \frac{1}{1-e^{-z}} \\
\frac{dy}{dz} &= \frac{d}{dz} \left(\frac{1}{1-e^{-z}}\right) \\
 &= -\frac{1}{(1-e^{-z})^2} \frac{d}{dz} (1 - e^{-z}) \\
 &= -\frac{1}{(1+e^{-z})^2}(-e^{-z}) \\
 &= \frac{e^{-z}}{(1+e^{-z})^2} \\
 &= \frac{1 + e^{-z} - 1}{(1+e^{-z})^2} \\
 &= \frac{1}{1 + e^{-z}} \left(\frac{1 + e^{-z} - 1}{1+e^{-z}}\right) \\
 &= \frac{1}{1+e^{-z}}\left(1-\frac{1}{1+e^{-z}}\right)=y(1-y)\\
Thus, \frac{dy}{dz} &=y(1-y)
\end{align*}
$$

**Advantages and Disadvantages**:   
The advantages of using sigmoid function as activation function are:
* It introduces **non-linearity** which enables the network to learn a complex mapping between input and output.
* It **prevents blowing of the activation** as it scales any input to value between 0 and 1.

where as, it suffers **sharp damping of gradient during the backpropagation** from deep hidden layers to the input layers, gradient saturation, and slow convergence i.e.   

* Gradient **vanishes** when the input is large or small.   
When x is large, then $y \rightarrow 1,$ implies $y^{'} \rightarrow 1(1-1)=0 $,
as shown in above graph of sigmoid and it's derivative.    

  Beside this, the range of it's derivative is $(0, 1/4]$. This means that  the value while **backprogating will be very small or at maximum** it can be  $1/4$ and hence, it will result in **vanishing gradient problem if backpropagated throuh deep neural network **and hence, the neural network will learn nothing.
* Optimization of network gets harder because the output isn't **zero centered(it is centered at 0.5).**

\




### **Hyperbolic Tan Activation Function**

**Advantages and Disadvantages**

This activation function is very similar to the sigmoid activation function **except the output is zero centered as the mapping scales the value between -1 and 1.** The tanh function couldn't solve the vanishing gradient problem suffered by sigmoid function. The main advantage is that it produce **zero centered output** thereby aiding the back propagation process. Optimization of the network using hyperbolic tan activation is easier.

The hyperbolic tan activation function is given as:
$$ f(x)=\frac{1-e^{-2x}}{1+e^{-2x}}$$

**Domain**: $- \infty \space to + \infty $   
**Range**: [$-1 \space to 1$]

<figcaption align="center">Figure 7: Tanh Function Plot </figcaption>

Hyperbolic tan activation is mostly used in **Recurrent Neural Network for natural language processing.**

Similarly, the derivative of tanh with respect to x (input values) is given by:

According to the quotient rule:

<center>
$\dfrac{d(f/g)}{dx}=\dfrac{gf^\prime-fg^\prime}{g^2}$

$set f= \sinh, g=\cosh $  to get, $\space\space\space$
$\dfrac{d\tanh}{dx}=\dfrac{\cosh\cdot\sinh^\prime-\sinh\cdot\cosh^\prime}{\cosh^2}$
</center>
Now,
<center>
$\sinh^\prime=\dfrac{1}{2}(e^x+e^{-x})=\cosh\\
\cosh^\prime=\dfrac{1}{2}(e^x-e^{-x})=\sinh$ 
</center>
Thus, 

<center>$\dfrac{d\tanh}{dx}=\dfrac{\cosh^2-\sinh^2}{\cosh^2}=1-\left(\dfrac{\sinh}{\cosh}\right)^2=1-\tanh^2$</center>

Now, 

<center>$\dfrac{1}{\cosh^2}=1-\tanh^2$</center>

Proof:
<center>$1-\dfrac{\sinh^2 x}{\cosh^2 x}= \dfrac{\cosh^2 x-\sinh^2 x}{\cosh^2 x}$

and

$\dfrac{1}{\cosh^2 x} = {\operatorname{sech}^2 x}$
</center>

Thus,

<center>$\boxed{\dfrac{d\tanh}{dx}=1-\tanh^2=\dfrac{1}{\cosh^2 x}}$</center>

###  **Rectified Linear Unit (ReLU)**
ReLU is an activation function famous for CNNs (Convolutional Neural Network) and most commonly used as activation in hidden layers in a neural network.

ReLU is defined as: 
$$R(z)=max(0,z)$$

<center>
<figure>
<img  src="https://docs.google.com/uc?export=download&id=1x_deKRGQS_5vP4WPekuq210psKJ1dA49" alt="Relu activation" height="384" width="473">
<figcaption align="center">Figure 8: ReLU Function Plot </figcaption>
</figure>
</center> 


The ReLU function acts as a **threshold to each input element where values less than zero are set to zero**. ReLU function rectifies the values of input less than zero thereby forcing them to zero and **eliminating the vanishing gradient problem.**

The derivative of ReLU activation function is:

\begin{equation} 
f'(x)=
    \begin{cases}
      1, & \text{if}\ x>0 \\
      0, & \text{otherwise}
    \end{cases}
\end{equation}


Here, The function is mostly differential but the value of the derivative of ReLU function is **undefined at** $x=0$. One way to deal with this problem while backpropagating is to assign an arbitrary value for derivative at $x=0$. Some common values are 0, 0.5 and 1.

**Advantages and Disadvantages**   
One major advantage of using ReLU is the **reduced likelihood of the gradient to vanish. In the region, $z>0$ the gradient has a constant value**. The constant gradient of ReLUs results in faster learning. Moreover, ReLU is computationally more efficient to compute than Sigmoid because Relu just needs to pick $max(0,z)$ and not perform expensive exponential operations as in Sigmoids.    
Also, **it introduces sparsity in the hidden units** as it squishes the value between zero to maximum.

However, ReLU has some problem too, i.e, it **tends to blow up activation (there is no mechanism to limit/rescale the output of the neuron, as "z" itself is the output)**. Relu function easily **overfits** compared to the sigmoid function. Moreover, there is another problem called **"Dying ReLU problem"** - if too many activations get below zero then most of the neurons in the network with ReLU will output zero, in other words, die and thereby prohibiting learning.

\

####  **Variations of ReLU**
#####  **Leaky ReLU**
Leaky ReLU is the specialized version of Randomised ReLU. When the value of parameter $\alpha$ is 0.01, then it becomes LReLU.

Mathematically, for Leaky ReLU,

$$ f(y)=
\begin{array}{cc}
  \{ & 
    \begin{array}{cc}
      \alpha y & y\lt 0 \\
      y & y\geq 0
    \end{array}
    \}&
\end{array}
$$


#####  **Sin ReLU (SReLU)**
Sin ReLU is defined as

$$ f(y)=
\begin{array}{cc}
  \{ & 
    \begin{array}{cc}
      Z & Z\gt 0 \\
      \varepsilon(sin(Z)-cos(Z)) & Z\leq 0
    \end{array}
    \}&
\end{array}
$$
where, $\varepsilon$ controls the wave amplitude.

#####  **Randomised ReLU (RReLU)**
Randomized ReLU is a **dynamic variant of leaky ReLU** where a random number sampled from uniform distrubution is used to train the network.
For RReLU, we have
$$ f(y)=
\begin{array}{cc}
  \{ & 
    \begin{array}{cc}
      \alpha y & y\lt 0 \\
      y & y\geq 0
    \end{array}
\end{array}
$$

Where,
$\alpha$ is a parameter, representing slope.

When $\alpha$=0.01, we have LReLU. However, when $\alpha \ne 0.01,$ then we obtain randomized ReLU.


##  Implementation and Visualisation

```
#Importing Numpy Library
import numpy as np
#Generating 200 values of x in between [-10,10]
x = np.linspace(-10, 10, 200)
print(x)
print(x.shape) # Validating the number of vlaues of x
```

```
#Defining Sigmoid Function
def sigmoid(x):
    return [(1/(1+np.exp(-i))) for i in x]

#Defining Hyperbolic Tan Function
def tanh(x):
    return [((1-np.exp(-2*i))/(1+np.exp(-2*i))) for i in x]
```

```
#Passing values of x to sigmoid function
y_sig=sigmoid(x)

#Passing values of x to tanh function
y_tanh=tanh(x)
```

```
#Importing Matplotlib Library
import matplotlib.pyplot as plt
%matplotlib inline

#Scatter Plot for Sigmoid Function
plt.scatter(x,y_sig, c='b')
#Scatter Plot for Tanh Function
plt.scatter(x,y_tanh, c="r")

#Labels
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

**In above plot between tanh and sigmoid, we can observe that the output for tanh is zero-centerd while sigmoid is not.**


###  Variants of ReLU 


```
#Defining Function for ReLU and its variants
def Relu(x):
    return [ (lambda i: i if (i>0) else 0)(i) for i in x]

def leaky_relu(x):
    return [ (lambda i: i if (i>=0) else 0.01*i)(i) for i in x]

def sin_relu(x):
    epsilon = 0.1
    return [ (lambda i: i if (i>0) else epsilon*(np.sin(i)-np.cos(i)))(i) for i in x]

def rand_relu(x):
    np.random.seed(1)
    alpha = np.random.rand(1)
    return [ (lambda i: i if (i>0) else alpha* i)(i) for i in x]
```

```
y_relu = Relu(x)
y_lrelu = leaky_relu(x)
y_srelu = sin_relu(x)
y_rrelu = rand_relu(x)

plt.scatter(x, y_relu, c='r')#ReLU=RED
plt.scatter(x, y_lrelu, c='g')#Leaky ReLU=GREEN
plt.scatter(x, y_srelu, c='b')#SReLU=Blue
plt.scatter(x, y_rrelu, c='y')#RReLU=Yellow

#Visualizing in a combine plot
plt.xlabel('x')
plt.ylabel('y')
plt.show() 
```

```
#Visualizations in multiple subplots
plt.figure(figsize=(10,10))
plt.subplot(321)
plt.tight_layout()
plt.scatter(x, y_relu, c='r')
plt.title('RELU')
plt.xlabel('x')
plt.ylabel('y')

plt.subplot(322)
plt.tight_layout()
plt.scatter(x, y_lrelu, c='g')
plt.title('Leaky Relu')
plt.xlabel('x')
plt.ylabel('y')

plt.subplot(323)
plt.tight_layout()
plt.scatter(x, y_srelu, c='b')
plt.title('Sin Relu')
plt.xlabel('x')
plt.ylabel('y')

plt.subplot(324)
plt.tight_layout()
plt.scatter(x, y_rrelu, c='y')
plt.title("Randomized Relu")
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

##  Use case of different activation functions

**Sigmoid function**: This activation function is used when we are doing Binary Classification modeling.

**Hyperbolic tan Function:**  Tanh function is mostly used in recurrent neural network for natural language processing and speech recognition task.

**ReLU Function**: This activation function eliminates the vanishing gradient problem. So, mostly used in the hidden units of deep neural network. Also, used in the output layer of the network which are used for object classification and speech recogition application.

## Key Takeaways
* In order to create a nonlinear decision boundary, activation function plays a vital role. Without activation functions, we can only learn linear decision boundaries.
* Sigmoid activation function is used when we are doing binary classification problems as it outputs a number between 0 and 1.
*  The output of tanh activation function is zero centered as the output is mapped between -1 and 1 which is its main advantage and aids the back propagation process. Optimization of the network using hyperbolic tan activation is easier.
* ReLU activation function eliminates the vanishing gradient problem. So, it is mostly used in the hidden units of deep neural network.

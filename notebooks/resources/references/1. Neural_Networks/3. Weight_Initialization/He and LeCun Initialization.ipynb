In the previous chapter, we discussed about Xavier Initialization which initialize the weights in network by drawing weights from the distribution with mean zero and a specific variance. Previously, sigmoid activation was widely accepted until ReLU activation came and dominate as a activation function. ReLU is popular because it solves the problem of vanishing and exploding gradients. As a result of this other initiaization techniques appeared which incorporate the same idea, that is balancing the variance between input and output. Different other initialization techniques are discussed as follows:

##He Normal Initialization



This technique is very similar to Xavier initialization and extends the same concept used by Xavier initializaiton when used with ReLU activation function. That mean for a network with sigmoid activation, Xavier initialization is used and for a network with ReLU activation, He initialization is used. The intuition behind this is: for a rectifying linear unit, half of it is not active. So half of the time it will produce 0. So we need to double the size of weight variance to keep the signal's variance constant. Therefore, we multiply our previous equation of variance (in Xavier initializatin, $Var(w_i)=\frac{1}{n_{in}}$) by 2.
Since, approximately half of the neurons are inactive so this reduce the number of neurons in input layer. So considering Xavier initialization and dividing the input neurons ($n_{in}$) by 2 we get,
$$Var(w_i)=\frac{1}{\frac{n_{in}}{2}}=\frac{2}{n_{in}}$$

This implies,
$$Std(\sigma)=\sqrt{Var}=\sqrt{\frac{2}{n_{in}}}$$

In this way, in He normal initialization, weights are initalized from Normal distribution with mean of 0 and a standard deviation of $\sqrt{\frac{2}{n_{in}}}$  i.e, $X\rightarrow N(0,\frac{2}{n_{in}})$

##LeCun Initialization

We have already discussed this initialization technique in Xavier initialization. Xavier initialization, actually, is the extension of Lecun initialization. Moreover, it is very similar with He normal initialization except, lecun initialization is used with sigmoid activation. Like He normal initialization it considers the neurons at the input layer and unlike Xavier initialization, it doesn't consider the neurons at the output layer.

The expression for variance in lecun initialization (input and output variance are similar) is already dervied in the chapter- Xavier initialization and is defined as:

$$Var(w_i)=\frac{1}{n_{in}}$$

The standard deviation is given as

$$Std(\sigma)=\sqrt{\frac{1}{n_{in}}}$$

So in Lecun initialization, weights are initialized from normal distribution with mean of 0 and a standard deviation of $\sqrt{\frac{1}{n_{in}}}$  i.e, $X\rightarrow N(0,\frac{1}{n_{in}})$



## Difference between Xavier, He and LeCun Initialization


>Xavier Initialization| He Initialization | Lecun Initialization | 
>---|---|:---
>Xavier Initialization is used with Sigmoid activation.|He Initialization is used with Relu activation|LeCun Initiaization works well with sigmoid activation
>Each randomly generated weight is multiplied by variance 2/(Nin + Nout).|Each randomly generated weights are multiplies with the variance 2/N |Each randomly generated weights are multiplies with the variance 1/N

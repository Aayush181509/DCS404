##Introduction to Weight Initialization


In order to train  and optimize neural network, initialization is the first concern to be considered. It is an initialization that determine how well the the successive training or learning process will proceed and the time it takes for training. Proper initialization is very crucial for faster convergence and better optimization of the network. 

Question may arise, why it actually matters on how we initialize. When we are working with deep neural networks, initializing the network with the right weights can be the difference between the network convergance in a reasonable amount of time and the network loss function not improving even after hundreds of iterations.

 Lets consider a deep neural network in which weights are initialized to a very small value. Their gradient tends to get smaller as we move backward through the layers. This cause neurons in earlier layers learn much slowly than neurons in later layers. As a result the weight update is negligible and this cause **Vanishing gradient problem**. If the gradients become 0 on the first few layers then it completely vanishes on the upcoming layers. Therefore, the early layers are the most important one and also called the building blocks. 

Similarly, if we initialize weights to a large value, then the gradients gets much larger in the earlier layers. This result in large weight updates causing  weights to explode to infinity(NaN). This is called **exploding gradient problem.**

The convergence of a network is dependent on how we update the parameters of the network. If the parameters are updating very slowly (caused by vanishing gradient) or fluctuated (caused by exploding gradient descent) then the convergence of network will be tedious and time consuming.

$$a_1=\sigma(0)$$

Now let's see the gradient

For,
$$\frac{dE}{dw_1}=\frac{dE}{da}\frac{da}{dz}\frac{dz}{dw_1}$$

The last term is,
$$\frac{dz}{dw_1}=a_1=0.5$$

The second term is, $$\frac{da}{dz}=a(a-1)=0.5(1-0.5)=0.25$$
The first term is,
$$\frac{dE}{da}=-(expected-output)=-(1-0.5)=-0.5$$


For,
$$\frac{dE}{dw_2}=\frac{dE}{da}\frac{da}{dz}\frac{dz}{dw_2}$$

The last term is,
$$\frac{dz}{dw_2}=a_2=0.5$$

The second term is, $$\frac{da}{dz}=a(a-1)=0.5(1-0.5)=0.25$$
The first term is,
$$\frac{dE}{da}=-(expected-output)=-(1-0.5)=-0.5$$
$assume$ $expected$=1

Since all the three terms are same in both case gradients of $w_1$ and $w_2$ are going to be same value, i.e $\frac{dE}{dw_1}=\frac{dE}{dw_2}$. Both of the weights will get updated with the same value and hence will be equal.

If we do the same process for other weights, we will get the same result. This means initialization by the same value no matter what that value is (e.g. 1, 2, 0.2) causes the weight update by the same amount. This problem is called **symmetry breaking problem**.

So when initialized all the weight by zero no features are able to forward through the network. Multiplying the features value with zero is just nullifying the input features. When initialized by any value cause  neurons to learn the same thing i.e, they are showing symmery nature.

So, whatever is the input, if the weights are same then units on the hidden layers will be same too. (Clearly explained on zeros initialisation above.)

Moreover, the same value weights initialisation techniques like: Zeros and Ones initialisation mainly affects the neural networks where each layer are fully connected with each other layer.

This is the main reason that brings out symmetry problem. Hence, the weights should be initialised randomly. Random initialisations are further discussed below.



### Random initialization

In order to avoid symmetry problem random initialization technique is used.
In this technique, the weights are initialized randomly following a certain distribution like uniform, normal, etc. This technique is better than initializing with the same value as neurons in the hidden layer will be able to learn different features.

Before moving forward lets discuss the different probability distributions.

#### **ii. Uniform distrbution**

The probability distribution function of uniform distribution is shown in figure below

This shows that the probability of occuring the value between a and b is equal i.e, $\frac{1}{b-a}$
Uniform distribution is defined as

$$ f(x) =
\begin{cases}
    \frac{1}{b-a} &\text{for } a \leq x\leq b\\
    0 &\text{otherwise}
\end{cases}
$$

Uniform distribution is represented as
$$X\sim u(a,b)$$

<br>

In random initialization, weights are initialized by selecting values randomly from either **uniform distribution** or **gaussian distribution**. The questions may arise which one to use. Well, this is unclear which one performs better over another. More or less it doesn't matter which one you choose. Moreover, for initialization data may be chosen from the specific range as shown above in  normal distribution curve. For example we may initialize weights from a normal distrubution with a mean of zero and a standard deviation of 1. Though random initialization works well, it may not be the best practise because the gradients might vanish or explode with just randomly thrown numbers. The problem may arise for a network with large number of neurons.
 
$$z=w_1x_1+w_2x_2+w_3x_3+.... x_{n-1}w_{n-1}+x_nw_n$$

For sigmoid activation,
$$g(z)=\frac{1}{1+e^{-z}}\approx1$$

For large neurons, the incremental contribution from all different terms in above equation cause the preactivation value(z) to grow large causing saturation and hence vanishing gradient problems. 

So what we learn from this part of material is that the number of neurons in a layer should be considered while defining the initialization of parameters. Also, the values initialised as weights in the neural networks are inversely proportional to the no. of neurons present in that layer.

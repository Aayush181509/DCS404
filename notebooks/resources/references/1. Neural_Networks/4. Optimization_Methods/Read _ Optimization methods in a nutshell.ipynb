## Comparison of SGD with momentum with SGD
SGD has trouble navigating ravines(areas where the surface curves much more steeply in one dimension than in another), which are common around local optima. So, at ravines, SGD oscillates across the slope and makes only slow progress along the bottom towards the local optimum. Momentum is a method that helps accelerate SGD in the relevant direction and dampens oscillations. Thus, with momentum, we gain faster convergence and reduced oscillation compared to Vanilla SGD.







## Comparison of Adagrad with SGD
Adagrad is an algorithm for gradient-based optimization that does just this: It adapts the learning rate to the parameters, performing smaller updates (i.e. low learning rates) for parameters associated with frequently occurring features, and larger updates (i.e. high learning rates) for parameters associated with infrequent features. For this reason, it is well-suited for dealing with sparse data.  Adagrad was found to greatly improve the robustness of SGD.
One of Adagrad's main benefits is that it eliminates the need to manually tune the learning rate.
Adagrad's main weakness is its accumulation of the squared gradients in the denominator. Since every added term is positive, the accumulated sum keeps growing during training. This in turn causes the learning rate to shrink and eventually become infinitesimally small, at which point the algorithm is no longer able to acquire additional knowledge.


## Comparison of RMSprop with SGD
RMSprop was developed to resolve Adagrad's radically diminishing learning rates. It was found to greatly improve the robustness of SGD. It is faster compared to Adagrad as it solves the vanishing gradient problem of Adagrad by using an exponentially weighted moving average in place of gradient accumulation.


## Comparison of ADAM with SGD
Adam can be looked at as a combination of RMSprop and Stochastic Gradient Descent with momentum. It uses the squared gradients to scale the learning rate like RMSprop and it takes advantage of momentum by using the moving average of the gradient instead of the gradient itself like SGD with momentum. Adam is great, it's much faster than SGD, the default hyperparameters usually works fine, but it has its own pitfall too. Many accused Adam has convergence problems that often SGD + momentum can converge better with longer training time.

## Choosing the Right Optimisation Algorithm

We discussed the series of related algorithms that each seek to address the challenge of optimising deep models by adapting rate for each model parameter.

At this point, an obvious question is: which algorithms should one use?

But it's unfortunate to say that there is no consensus at this point. While some results from different researchers suggest that the family of algorithms with adaptive learning rates performed fairly robust.

Currently, the most popular optimisation algorithms actively in use include SGD, SGD with momentum, RMSprop, Adagrad and Adam.


If our input data is sparse, then we likely achieve the best results using one of the adaptive learning-rate methods. An additional benefit is that we won't need to tune the learning rate but likely achieve the best results with the default value.

In summary, RMSprop is an extension of Adagrad that deals with its radically diminishing learning rates. Adam adds bias-correction and momentum to RMSprop. RMSprop and Adam are very similar algorithms that do well in similar circumstances. It is shown by some researchers that its bias-correction helps Adam slightly outperform RMSprop towards the end of optimization as gradients become sparser. Adam might be the best overall choice.

## Introduction
As we have seen in the earlier chapter that Adagrad's main weakness is its **accumulation of the squared gradients in the denominator**. Since every added term is positive, **the accumulated sum keeps growing** during training. This in turn **causes the learning rate to shrink and eventually become infinitesimally small, at which point the algorithm is no longer able to acquire additional knowledge**.
Following the disadvantage of Adagrad, we are here to eliminate this problem. **RMSprop modifies Adagrad to perform better in the nonconvex setting by changing the gradient accumulation into an exponentially weighted moving average**. Basically, we have to prevent this situation where $\alpha_t$ increases to a higher number. We should restrict $\alpha_t$ from becoming a high number. So here, in this algorithm we will make a slight change in the earlier formula of Adagrad optimisation technique.

Here in RMSprop,

$$V_{t} = \beta V_{t-1} + (1-\beta)[\frac{\partial Loss}{\partial w_{t}}]^2$$

$$ {\alpha}' = \frac{\alpha}{\sqrt{V_{t}+\epsilon }}$$

As we see here, instead of taking the sum of the square of the derivatives we are doing the exponentially weighted average. Doing this, it's sure that **$V_{t}$ will increase but not as much as $\alpha_{t}$. It restricts on the increment due to which learning rate decreases slowly and convergence will not take long time.**

Empirically, RMSprop has been shown to be an effective and practical optimisation algorithm for deep neural networks. It is currently one of the go-to optimisation methods been employed routinely by deep learning practitioners.

## Concept of momentum

This algorithm uses the concept of momentum, and takes it one step further. 

Recall the equation that determines momentum using exponentially weighted averages:

$$V_t = \beta \times (V_{t-1}) + (1-\beta)dW_{t}$$
Here, $dW_t$ is the current value for the gradient, $\beta$ is the hyperparameter controlling how many previous values are taken into consideration. The usual value for $\beta$ is 0.9. $V_{t-1}$ is the previous value derived in this way. If we are computing the first value, and we do not have a previous value, we just put zero. 

When we update the gradients, we use this value, instead of using the values of the gradients directly. 
$$W_{t+1} = W_{t} - \alpha V_t$$
where $\alpha$ is the learning rate. Since the unwanted vibrations orthogonal to the direction of the minimum will be dampened out by using momentum, we will be able to use a larger learning rate as well. 

## Taking it a step further

RMSprop stands for **Root Mean Squared propagation**. This algorithm uses a variant of this equation. 
$$V_t = \beta \times (V_{t-1}) + (1-\beta)dW_{t}^2$$

**Notice the square on the (element-wise) value of the gradient. This square makes the values of the gradients larger, and thus is better at dampening the unwanted vibrations**. 

When updating the weights we use the following equation
$$W_{t+1}= W_t - \alpha \frac{dW_t}{\sqrt[]{V_t}}$$

As you can see, it is using the true value of the gradient divided by the root of the exponentially weighted average of the squares of the gradient. 

Practically speaking, we want to prevent dividing by zero. Sometimes, when the gradients are small, or when the numbers are rounded up, we might end up with the term $\sqrt[]{V_t}$ being zero. To prevent this, we put a very small number $\epsilon$ inside the root. So the equation becomes

$$W_{t+1}= W_t - \alpha \frac{dW_t}{\sqrt[]{V_t+\epsilon}}$$

Usually the value for $\epsilon$ is an arbitrarily small number like $10^{-8}$.

**Remember that even though we've just been talking about applying momentum to the weights, but the same applies to the bias term as well**. 

$$S_t = \beta \times (S_{t-1}) + (1-\beta)db_{t}^2$$
$$b_{t+1}= b_t - \alpha \frac{db_t}{\sqrt[]{S_t+\epsilon}}$$

Here, $db$ is the current gradient for the bias term. $S_t$ is the exponentially weighted average. 

## Comparison of RMSprop with SGD

RMSprop was developed to resolve Adagrad's radically diminishing learning rates. It was found to greatly **improve the robustness of SGD**. It is **faster compared to Adagrad** as it solves the vanishing gradient problem of Adagrad by using an exponentially weighted moving average in place of gradient accumulation.

## Key Takeaways

1.  RMSprop modifies Adagrad to perform better in the nonconvex setting by changing the gradient accumulation into an exponentially weighted moving average.
2. $V_t$ increases but not as much as $ \alpha_t\ $. So, due to restriction on the increment of $V_t$,  learning rate decreases slowly and convergence will not take long time.

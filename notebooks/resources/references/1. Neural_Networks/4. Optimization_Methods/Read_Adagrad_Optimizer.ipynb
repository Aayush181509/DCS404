## Introduction
Adagrad also called as Adaptive Gradient Algorithm is an algorithm for **gradient based optimisation which adapts the learning rate to the parameters, performing smaller updates**.

There are hyperparameters used in all algorithms, out of which learning rate is one of the most sensitive ones. Hyperparameter setting is a daunting task. If we set it too small, parameter update will be very slow and it will take very long time to achieve an acceptable loss and on the other side, if we set it too large , then the parameter will move all over the function. One common way to mitigate this problem is to **choose different learning rate for different dimension**. But in case of deep neural networks, where we can have millions of dimensions it's manually impossible to tune the learning rate. So here, we use one of the earlier algorithms called AdaGrad.

**This algorithm adaptively scales the learning rate for each dimension.**

Dataset may have different nature.We perform larger updates for infrequent parameters and smaller updates for frequent parameters.
So, let's see how does Adagrad manages this.




## Mathematics behind Adagrad

For SGD, momentum, NAG we have,

$$w_t = w_{t-1}-\eta\frac{\partial Loss}{\partial w_{t-1}}$$

In case of Adagrad,
$$w_t = w_{t-1}-{\eta}'\frac{\partial Loss}{\partial w_{t-1}}$$
where, $$ {\eta}' = \frac{\eta}{\sqrt{\alpha_t+\epsilon }}$$
$\epsilon$ is kept to avoid the division by 0. It is always a small positive number.
$$ \alpha_t = \sum_{i=1}^{t}[\frac{\partial Loss}{\partial w_i}]^{2}$$

As the iteration goes on, **value of $\alpha_t $ increases due to which learning rate decreases and weight update also decreases accordingly. So the algorithm will be converging slowly and nicely.**

Hence, we see here that the value of learning rate is changing with respect to features, layers, iterations.

## Comparison of Adagrad with SGD

Adagrad is an algorithm for gradient-based optimization that does just this: **it adapts the learning rate to the parameters**, performing smaller updates (i.e. low learning rates) for parameters associated with frequently occurring features, and larger updates (i.e. high learning rates) for parameters associated with infrequent features. For this reason, it is well-suited for dealing with sparse data. Adagrad was found to greatly improve the robustness of SGD. One of Adagrad's main benefits is that **it eliminates the need to manually tune the learning rate**. Adagrad's main **weakness is its accumulation of the squared gradients in the denominator**. Since every added term is positive, the accumulated sum keeps growing during training. This in turn **causes the learning rate to shrink and eventually become infinitesimally small, at which point the algorithm is no longer able to acquire additional knowledge**.

We will fix this problem in the next optimisation algorithm called RMSprop.


## Key Takeaways

1. Adagrad Algorithm is an algorithm for gradient based optimisation which adapts the learning rate to the parameters, performing smaller updates.
2. Sometimes $\alpha_t $ becomes a very high number as the iteration increases. It causes learning rate to be too small and there is no learning anymore. This problem is solved by RMSprop.

## Layer Normalization
Layer normalization is developed by Geoffery Hinton. Layer normalization is very similar to batch normalization. In batch normalization, we normalize each feature (each neuron) in a layer according to the batch of samples.
For example, in batch normalization, input values of the same neuron from different images in one mini-batch are normalized.

However, in layer normalization, we normalize each feature (each neuron) in a layer according to the features in that layer. So, mean and standard deviation are calculated across the features of a layer (across the output of a layer). For example, in layer normalization, input values for different neurons in the same layer are normalized without consideration of mini-batch.

As shown in the above figure, we normalize the output across the layer output.
So, batch normalization and layer normalization differ in the way we compute the mean and standard deviation.

We first compute the mean and variance of output of neurons in a layer. After that, we normalize the output of neurons and then allow the network to learn the scaling and shifting parameters that are best optimal.

Now let's see the mathematics involved.

 $$ y_i=LN_{\gamma,\beta}(x_i)  $$

$$\mu_{B}=\frac{1}{m}\sum_{i=1}^{m}x_i$$   
$$\sigma_B^2=\frac{1}{m}\sum_{i=1}^m(x_i-\mu_B)^2$$  
$$\hat{x_i}=\frac{x_i-\mu_B}{\sqrt{\sigma_B^2+\epsilon}}$$

$$y_i=\gamma\hat{x_i}+\beta \equiv LN_{\gamma,\beta}(x_i)$$

where,

$x_i:$ Output at the neuron 'i' in a layer.

$m: $ The number of neurons in a layer.

$\mu_B:$ Mean over all neurons in a layer.

$\sigma_B^2:$ Variance over all the neurons in a layer. 

$\hat{x_i}:$ Normalized value.

$\gamma, \beta:$ Scaling and shifting parameters.

$\epsilon : $ Very small number to prevent division by zero.

**Benefits of Layer Normalization**
- Normalization parameters depend only on the weighted summed inputs to layer at current time-step and same gain$(\gamma)$ and bias$(\beta)$ parameters can be shared in all time-steps in RNN. This allowed options for taking advantage of normalization in RNN
- No need for memorizing and calculating training population statistics for normalization during inference.

**Limitation of Layer Normalization**
- Performance of layer normalization in case of CNN is found to be poorer than batch normalization


# Takeaways
The main takeaways from the lesson are:
- In layer normalization, input values for different neurons in the same layer are normalized without consideration of mini-batch.
- Mean and variance are computed on all neurons of a layer.

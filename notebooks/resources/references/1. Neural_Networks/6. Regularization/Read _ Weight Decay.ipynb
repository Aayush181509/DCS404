## Introduction

After L1 and L2 regularization, we are going to learn about a new concept called **Weight Decay**. It is one of the regularization technique but related to optimization. Similar to the regularization techniques we discussed before, weight decay prevents the model from overfitting by penalizing the weights.


## Why weight decay?

You might be wondering why use weight decay while we already have $L_1$ and $L_2$ regularisation. The fact is that weight decay is an additional term that causes the weights to exponentially decay to zero or near to zero. At the end of each batch update, weights are multiplyed by a number. That number is called weight decay value.


The effect of weight decay might be similar to L2 regularisations in optimizers without momentum such as SGD, but different effects can be seen for optimizers with momentum like adam, RMSprop etc which we will be discussing later in this chapter.


## Working of weight decay

For a model with weights '$w$', learning rate '$\alpha$' and the mean squared loss '$L_e$', we know that the weight update rule is written as:

$$
w = w - \alpha \frac{\partial L_e}{\partial w}
$$

Adding a term called weight decay to the above cost function gives us:

$$
L_e = L_e + \lambda \frac{w^2}{2}
$$

Differentiating the above equation with respect to '$w$' we get:

$$
\frac{\partial L_e}{\partial w} = \frac{\partial L_e}{\partial w} + \frac{\lambda}{2}  \frac{\partial (w^2)}{\partial w}
$$

After carrying out the necessary steps we end with:

$$\frac{\partial L_e}{\partial w} = \frac{\partial L_e}{\partial w} + {\lambda} {w}
$$

Substituting the above equation to the weight update rule we get:

$$
w = w - \alpha ( \frac{\partial L_e}{\partial w} + {\lambda} {w} )
$$

In the above relation, '$\lambda $' is the weight decay value.


**Numerical Example**

For a model with weights $w_0 = 0.15,$ $w_1 =0.5$, $w_2 = 0.25$; learning rate $\alpha = 0.01$; $\frac{\partial L}{\partial w_0} = 0.2425, \frac{\partial L}{\partial w_1} = 0.1212,  \frac{\partial L}{\partial w_2} = 0.25$ and the mean squared loss $L_e=0.533$, weight update rule **without weight decay** can be written as:
$$
w_0 = 0.15 - 0.01 * 0.2425 = 0.147575
$$
$$
w_1 = 0.5 - 0.01 * 0.1212 = 0.498788
$$
$$
w_2 = 0.25 - 0.01 * 0.0606 = 0.249394
$$


**With weight decay**(Assuming weight decay value $\lambda = 0.1$)
$$ 
w_0 = 0.15 - 0.01*(0.2425 + 0.1*0.15) = 0.147425
$$
$$
w_1 = 0.5 - 0.01*(0.1212 + 0.1*0.5) = 0.498288
$$
$$
w_2 = 0.25 - 0.01*(0.0606 + 0.1*0.25) = 0.249144
$$


Thus from the above numerical example we can see that weight decay penalizes large weights and prevents overfitting by acting as a regularizer.


## Weight decay in SGD

In the above relation we see that, when we are adding weight decay while modifying the weight update directly, it acts just like $L_2$ regularisation. Above relation was derived on Vanilla SGD or simply SGD. hence we can see that $L_2$ regularisation and the weight decay seems same. Here $L_2$ equals regularisation.
But, for the case of when we add momentum or use Adam, the regularization and the weight decay aren't same. And $L_2$ does not equals regularization.


## Weight decay in Adam
<!--{clarify dissimilarities between L2 regularisation in Adam and weight decay in Adam along with code}-->

As we know that Adam optimizer is equipped with momentum. And as discussed earlier, weight decay shows significant effect on optimizers with momentum. It regularizes variables with large gradients more than L2 regularization would, which was proven to yield better training loss.

When using L2 regularization in Adam the gradients aren’t subtracted from the weights directly. First we compute a moving average.The penalty we use for large weights gets scaled by moving average of the past and current squared gradients and therefore weights with large typical gradient magnitude are regularized by a smaller relative amount than other weights. 

In contrast, weight decay regularizes all weights by the same factor as shown in the following equation:
$$ 
  w_t = w_{t-1} - \eta(\frac{\hat{m}}{\sqrt{v_t} + \epsilon} + \lambda w_{t-1})
$$


## When use Weight Decay?
As mentioned earlier, unlike L1 and L2 regularization, weight decay causes weight to exponentially decay to zero or near to zero. Hence weight decay helps to generalize the data by converting the weights near to zero. The effect of weight decay is best seen in the optimizers with momentum. It is because, increase in momentum increases the decay rate of the weight.

Hence it is safe to say that, we can use weight decay whenever we use optimizers with momentum.


## Some observed examples
Weight decay has different regularisation effect depending of algorithm and architecture.
In the paper “THREE MECHANISMS OF WEIGHT DECAY REGULARIZATION” we can observe some examples:

1. In first-order optimisers like SGD and adam with batch normalisation, weight decay  acts by way of the effective learning rate where regularisation effect of gradient noise is increased.
2.  In Kronecker-Factored Approximate Curvature (KFAC), weight decay is equivalent to regularizing the squared Frobenius norm of the input-output Jacobian.
3. In KFAC, without weight decay, weights become large and effective damping term increases causing the KFAC to turn into a first-order optimizer. But weight decay keeps the effective damping term small and hence KFAC retains its second-order properties improving generalisation.

You can learn more about these from [THREE MECHANISMS OF WEIGHT DECAY REGULARIZATION](https://openreview.net/pdf?id=B1lz-3Rct7).




## Key Takeaways
* Weight decay is an additional term that causes the weights to exponentially decay to zero or near to zero.
* The effect of weight decay is different from L2 regularisations in optimizers like adam, RMSprop, unlike for SGD without momentum.

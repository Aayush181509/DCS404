The task that Shila performed is known as clustering. To formally put it, clustering is the process of partitioning the observations into groups or clusters such that observations in the same cluster tend to be similar, and those in the different clusters tend to be dissimilar.


### Clustering in Machine Learning

Clustering is an unsupervised learning algorithm that deals with the identification of subgroups in the data such that data points in the same subgroup have high similarity, while data points in different clusters have a low similarity.

### Clustering is an Unsupervised Learning

Clustering seems similar to classification. In both classification and clustering, we categorize a given data point as belonging to some group/class. But unlike classification where we have the true label to guide the model training process, in clustering we do not have the true label to guide or supervise the training. Therefore, clustering is an unsupervised machine learning method.


### Types of Clustering

There are two categories of clustering algorithms based on their cluster assignment. They are explain below:


#### Hard Clustering

Hard clustering algorithms assign each point to one cluster definitively. That is, each point belongs completely to one and only cluster. From Figure 2(b), you can see that one data point has only one color; it means it belongs to only one cluster. Hence, that is the example of a hard clustering.


#### Soft Clustering

On the other hand, soft clustering algorithms may assign each point to more than one cluster. If you observe the data points carefully in Figure 2(b), you may wonder why some points that seem to be in-between the two clusters assigned to only one of them. In that case, we may want to assign those points in two or more clusters, making it a soft clustering algorithm.

Also, from Figure 2, if there are two distinct clusters formed by land animals and aquatic animals in our dataset, crocodiles may be 'soft-assigned' to both of them. With hard clustering, we would have had to definitely assign crocodiles to one of these two clusters, which doesn't make a whole lot of sense.


### Types of Clustering Algorithms

Clustering algorithm based on their clustering approaches can be divided into following categories:

1. **Partitioning-Based Clustering:**  
Partitioning based clustering algorithms work by subdividing the data sets into some number of clusters (_K_). The clusters are generally identified by the centroid points such as the mean. A data point belongs to a cluster whose centroid is closest to it, i.e. centroid with minimum distance. With partition based clustering, the data shown in the Figure 2(a) can be partitioned as,

2. **Hierarchical Clustering:**   
Hierarchical clustering is a connectivity-based clustering algorithm where either a bigger cluster is formed by connecting two similar smaller clusters, or two smaller clusters are formed by breaking a bigger cluster. This type of clustering creates a hierarchy of connectivity, hence called hierarchical clustering. Figure below shows a simple dataset where hierarchical clustering is applied.

    </figure>
</div>

3. **Density-Based Clustering:**   
In density-based clustering, a region of densely connected data points creates clusters. If there is more than one densely connected region, then there is more than one cluster. Density-based clustering does not assume the shape of the clusters. Hence, they can be of any form but separated by a sparse region.

4. **Distribution-Based Clustering:**   
The goal of distribution based clustering is to assign the probability to the data points belonging to some probability distribution.

5. **Grid-Based Clustering:**   
Grid-based clustering divides the data space into a discrete space called grids. Then, the density of the data in each grid is used to classify it as noise or cluster. The clusters are then grouped based on neighboring locations to form the region of clusters and empty regions.

## Applications

Clustering is a popular unsupervised machine learning algorithm. Some of its applications are listed below:

* **Customer segmentation:** Customers can be clustered based on their purchases, location, activity. This helps sellers to know their customers and optimize the products accordingly. They can recommend new items to a specific type of customer and make marketing effective.

* **Anomaly detection:** Any observation with a low probability of occurring is called Anomaly (or outlier). In banking transactions, the activity of the customer can be analyzed and monitored for unusual activity. The unusual activity can then be looked at carefully to find fraud.

* **Data analysis:** A new dataset is easy to process if it is divided into different chunks with similar data. Each chunk can be analyzed separately and more effectively.

* **Image Segmentation:** Image pixels can be clustered and replaced with the mean value (centroid). It allows all colors to be represented by the number of clusters. It is useful in image segmentation as similar colors group together in one (mean) color.


## Key take-aways

The key take-aways from this chapter are:

- Clustering means grouping data points so that similar data belong to the same group and dissimilar data to different groups.

- Clustering is an unsupervised learning. It means we do not need the actual labels to group the data points.

- If a clustering algorithm assigns a data point to only one cluster, it is hard clustering. If it assigns data to more than one cluster, it is soft clustering.

- There are many different clustering algorithms with a lot of applications.
